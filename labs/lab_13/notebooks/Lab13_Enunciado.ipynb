**Docker** es un conjunto de programas que ejecuta aplicaciones dentro de **contenedores**: entornos aislados que traen su propio Python y sus propias librerías, así que la aplicación corre igual en cualquier computador con Docker y la misma arquitectura de procesador (lo veremos en 4.3). Dos ideas bastan por ahora. Una **imagen** es el paquete, de solo lectura: un sistema de archivos mínimo con Python, las librerías y el código de la aplicación.   <strong style="display:block; margin-bottom:8px;">📖 Docker: un demonio y un cliente</strong>

Cuando decimos «Docker» nos referimos a dos programas que conversan. El **demonio** (`dockerd`) es el motor: construye imágenes y ejecuta contenedores, y corre en segundo plano. El **cliente** (`docker`) es el comando que ustedes escriben en la terminal: no hace el trabajo, solo le envía órdenes al demonio y muestra lo que este responde.

En Windows y macOS ambos vienen juntos en **Docker Desktop**, que además trae una máquina virtual Linux mínima donde corren los contenedores. En Linux se instala **Docker Engine**, que corre directamente en el sistema. Que haya dos programas explica la falla más frecuente: si el demonio no está en ejecución, el cliente no tiene con quién hablar.
</div>

  <strong style="display:block; margin-bottom:8px;">📖 Artefacto de modelo y metadatos</strong>

Un modelo guardado solo como archivo no dice con qué datos se entrenó, con qué versión de scikit-learn ni qué columnas espera. Si el servicio de potabilidad empieza a responder distinto, nadie puede averiguar si cambió el modelo, el dato de entrenamiento o la biblioteca.

Un **artefacto de modelo** resuelve eso: el archivo con el modelo ya entrenado **más sus metadatos**, que dicen cómo se produjo. Los que usaremos son la fecha de entrenamiento, la versión de scikit-learn, el orden de las variables de entrada, los hiperparámetros, la semilla, las métricas de validación y una **huella** de los datos: un resumen corto del contenido del CSV (aquí, los primeros 16 caracteres de un SHA-256) que cambia si cambia un solo byte.

Vive en una carpeta con dos archivos:

```text
models/
├── model.joblib
└── metadata.json
```

Los metadatos no sirven para elegir entre modelos: sirven para que el servicio compruebe, antes de atender a nadie, que el archivo calza con el contrato, es decir, que las variables vienen en el orden esperado y que la versión de scikit-learn es la misma con que se entrenó.
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Serializar un modelo: `pickle` y `joblib`</strong>

Para guardar un modelo en disco hay que convertir el objeto de Python en bytes, lo que se llama **serializar**. `joblib`, la biblioteca que scikit-learn recomienda para guardar modelos, usa por debajo `pickle`, el mecanismo estándar de Python. Al **cargar** el archivo, `pickle` reconstruye el objeto siguiendo las instrucciones que el propio archivo trae, y esas instrucciones pueden llamar a cualquier función de Python.
</div>

<!-- WARNING -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #d29922; background:rgba(210,153,34,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">⚠️ Cargar un pickle ejecuta código</strong>

Un archivo `pickle` no es un dato inerte: es una receta de reconstrucción, y quien la escribe elige qué funciones se llaman. La clase de abajo le indica a `pickle` que, para reconstruirla, llame a `print`. Un archivo malicioso podría pedir `os.system` con el comando que quisiera, y se ejecutaría con los permisos de quien lo carga, solo por abrirlo:
</div>

````python
import pickle


class Inofensivo:
    # Para reconstruir el objeto, pickle llama a lo que devuelva __reduce__.
    def __reduce__(self):
        return (print, ("Este código se ejecutó solo por cargar el archivo.",))


contenido = pickle.dumps(Inofensivo())
_ = pickle.loads(contenido)
````

Por eso un modelo se carga **solo** desde una fuente de confianza, como la carpeta que produce el propio laboratorio. Hay una segunda falla, más silenciosa: scikit-learn emite un `InconsistentVersionWarning` cuando un archivo guardado con una versión se carga con otra, y no garantiza que las predicciones sean idénticas. De ahí que los metadatos registren la versión de scikit-learn y que el servicio la compare antes de atender a nadie.

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Contrato de entrada</strong>

Un servicio recibe datos de quien no controla: un sensor que falla, una persona que se equivoca al escribir. Si esos datos llegan sin revisión hasta el modelo, este responde igual y no avisa que la entrada no tenía sentido. Un **contrato de entrada** declara qué datos acepta el servicio (nombres, tipos y rangos) y rechaza el resto antes de que lleguen al modelo.

Declarar los tipos no basta, porque Python **no los hace cumplir**: las anotaciones son información para quien lee el código y para las herramientas, y al ejecutar se ignoran. En la panadería de don Abel:
</div>

````python
def pedir_pan(kilos: float) -> float:
    return kilos * 2


print(pedir_pan(3))
print(pedir_pan("hola"))
print(pedir_pan(-5))
````

Python aceptó un texto donde la anotación decía `float` y devolvió `holahola`, y también aceptó pedir menos cinco kilos de pan. Hace falta una biblioteca que **verifique** el contrato cada vez que llega un dato.

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Validación con Pydantic</strong>

**Pydantic** verifica datos a partir de anotaciones de tipo. Los tipos básicos que usaremos son `int` (enteros), `float` (decimales), `str` (texto), `bool` (verdadero o falso) y `None` para un valor ausente: `float | None` significa «un decimal o nada».

Un modelo de Pydantic es una **clase** que hereda de `BaseModel`. Una clase reúne en un solo lugar el nombre, el tipo, el rango y la obligatoriedad de cada campo, y se reutiliza donde llegue un dato: FastAPI, en la etapa 2, la usa para validar el cuerpo de cada petición sin código adicional. Al crear un objeto de esa clase, Pydantic **valida**: convierte lo que puede (el texto `"3"` pasa a `3.0`), rechaza lo que no y lanza un `ValidationError` que dice qué campo falló y por qué.

`Field` agrega restricciones a un campo: `gt` (mayor que), `ge` (mayor o igual que), `lt` (menor que) y `le` (menor o igual que). Un campo sin valor por defecto es obligatorio.
</div>

````python
from pydantic import BaseModel, Field, ValidationError


class PedidoPan(BaseModel):
    kilos: float = Field(gt=0, le=200)
    tipo: str
    nota: str | None = None  # opcional: puede faltar


print(PedidoPan(kilos="3", tipo="marraqueta"))

rechazados = [
    {"kilos": -5, "tipo": "marraqueta"},
    {"kilos": "mucho", "tipo": "marraqueta"},
    {"kilos": 3},
]
for datos in rechazados:
    try:
        PedidoPan(**datos)
    except ValidationError as error:
        detalle = error.errors()[0]
        print(detalle["loc"], "→", detalle["msg"])
````

Con tres entradas distintas, Pydantic nombró el campo y la razón de cada rechazo, y aceptó convertir `"3"` en `3.0`. En 1.2 declararán `MedicionAgua` con esta misma sintaxis.

  <strong style="display:block; margin-bottom:8px;">📖 API y URI</strong>

Una **API** (interfaz de programación de aplicaciones) es el conjunto de preguntas que un programa acepta de otro. Así como una persona le pide algo a la sanitaria con un formulario en pantalla, un programa se lo pide con una API. Cada pregunta apunta a un recurso mediante una **URI**, una dirección con partes que tienen nombre. Tomemos como ejemplo la de una biblioteca municipal:

```text
https://biblioteca.example.cl:443/libros/1?formato=json
```

| Parte | En el ejemplo | Qué indica |
| --- | --- | --- |
| Esquema | `https` | El protocolo. `https` viaja cifrado; `http`, no. |
| Host | `biblioteca.example.cl` | El computador que responde. En local, `localhost` es su propio computador. |
| Puerto | `443` | La «puerta» del computador por donde atiende el programa. Se omite cuando es la habitual (80 en `http`, 443 en `https`); nuestra API local usa la `8000`. |
| Ruta | `/libros/1` | Qué recurso se pide. |
| Parámetros | `?formato=json` | Opciones de la pregunta, como pares `clave=valor` separados por `&`. |
</div>

  <strong style="display:block; margin-bottom:8px;">📖 El protocolo HTTP</strong>

**HTTP** es el protocolo con que el cliente y el servidor se envían mensajes de texto: el cliente manda una **petición** y el servidor devuelve una **respuesta**. La petición lleva un **método** (qué quiere hacer), la ruta, **cabeceras** (datos sobre la petición, como la clave de acceso) y, a veces, un **cuerpo** con los datos. Así se ve una petición para agregar un libro y su respuesta:

```text
POST /libros HTTP/1.1
Host: biblioteca.example.cl
Content-Type: application/json
X-API-Key: clave-de-ejemplo

{"titulo": "Veinte poemas de amor y una canción desesperada"}
```

```text
HTTP/1.1 201 Created
Content-Type: application/json

{"id": 2}
```

Los métodos más usados son estos:

| Método | Para qué sirve | En la biblioteca |
| --- | --- | --- |
| `GET` | Leer, sin modificar nada. | `GET /libros/1` entrega el libro 1. |
| `POST` | Enviar datos para crear algo o procesarlos. | `POST /libros` agrega un libro. En nuestro servicio, `POST /predict` envía una medición. |
| `PUT` | Reemplazar un recurso completo. | `PUT /libros/1` cambia todos los datos del libro 1. |
| `PATCH` | Modificar solo una parte. | `PATCH /libros/1` cambia únicamente el título. |
| `DELETE` | Eliminar. | `DELETE /libros/1` borra el libro 1. |

En este laboratorio usaremos `GET` y `POST`.

  <strong style="display:block; margin-bottom:8px;">📖 Códigos de respuesta</strong>

La respuesta lleva un **código de estado** que dice qué pasó antes de mirar el contenido. Se agrupan por su primer dígito: `2xx` es éxito, `4xx` es una falla del cliente y `5xx` es una falla del servidor.

| Código | Significa | Quién falló |
| --- | --- | --- |
| `200` | Todo bien. | nadie |
| `201` | Se creó el recurso pedido. | nadie |
| `401` | La petición no se identificó correctamente (falta la clave o es inválida). | el cliente |
| `403` | El cliente se identificó, pero no tiene permiso. | el cliente |
| `404` | Lo solicitado no existe. | el cliente |
| `422` | El servidor entendió la petición, pero sus datos no son válidos. | el cliente |
| `500` | Algo se rompió adentro. | el servidor |

La distinción entre 4xx y 5xx importa: un 4xx le dice al cliente que **corrija su petición**; un 5xx, que **reintente más tarde o avise** a quien administra el servicio.

  <strong style="display:block; margin-bottom:8px;">📖 FastAPI</strong>

**FastAPI** es la biblioteca de Python con la que escribiremos la API. Una ruta se declara con un **decorador**, que asocia un método y una ruta con una función: `@app.get("/libros/{libro_id}")` hace que la función responda a los `GET` de esa ruta, y `{libro_id}` es una parte variable de ella.

Lo que la función declara con tipos, FastAPI lo valida con Pydantic: si la función pide `libro_id: int` y llega `abc`, responde un 422 sin que escribamos código para eso, y lo mismo ocurre si el cuerpo es un modelo de Pydantic. FastAPI además genera sola una página de documentación en `/docs` y trae `TestClient`, que prueba una API sin levantar un servidor. En la biblioteca:
</div>

````python
from fastapi import FastAPI, HTTPException
from fastapi.testclient import TestClient

biblioteca = FastAPI()
LIBROS = {1: "Veinte poemas de amor y una canción desesperada"}


@biblioteca.get("/libros/{libro_id}")
def obtener_libro(libro_id: int):
    if libro_id not in LIBROS:
        raise HTTPException(status_code=404, detail="El libro no existe.")
    return {"id": libro_id, "titulo": LIBROS[libro_id]}


cliente_demo = TestClient(biblioteca)
for ruta in ["/libros/1", "/libros/99", "/libros/abc"]:
    print(ruta, "→", cliente_demo.get(ruta).status_code)
````

`/libros/99` es un 404 porque el libro no existe; `/libros/abc` es un 422 porque `abc` no es un entero. En ninguno de los dos casos hay un error del servidor. Si al importar `TestClient` aparece un `StarletteDeprecationWarning` sobre `httpx`, es un aviso de Starlette acerca de su cliente de pruebas; no cambia ningún resultado de este laboratorio.

**`pydantic-settings`** aplica la idea de Pydantic a la configuración: una clase que hereda de `BaseSettings` declara cada variable como un campo con su tipo, y al crear el objeto la lee del entorno (y de un archivo `.env`, si se le indica), la convierte al tipo declarado y lanza un `ValidationError` si falta una obligatoria. Así la configuración queda en un solo lugar, con tipos, y cualquier módulo puede importarla.

`SecretStr` es el tipo para un secreto: al imprimir el objeto muestra `**********` en vez del valor, y para usarlo hay que pedirlo con `.get_secret_value()`. En la panadería:

````python
import os
from unittest.mock import patch

from pydantic import SecretStr
from pydantic_settings import BaseSettings, SettingsConfigDict


class ConfigPanaderia(BaseSettings):
    model_config = SettingsConfigDict(env_prefix="PAN_")

    clave_caja: SecretStr
    precio_base: float = 1500


entorno_demo = {"PAN_CLAVE_CAJA": "pan-caliente-123", "PAN_PRECIO_BASE": "1800"}
with patch.dict(os.environ, entorno_demo):
    config = ConfigPanaderia()

print(config)
print(config.precio_base, config.clave_caja.get_secret_value())
````

  <strong style="display:block; margin-bottom:8px;">📖 Imagen, contenedor, capa y volumen</strong>

Recordemos que una **imagen** es el paquete y un **contenedor**, esa imagen en ejecución. A diferencia de una máquina virtual, un contenedor comparte el núcleo de Linux del equipo que ejecuta Docker (el *anfitrión*; en Windows y macOS, el de la máquina virtual de Docker Desktop) y arranca en segundos.

Una imagen se construye como una pila de **capas**: cada instrucción del `Dockerfile` (`FROM`, `COPY`, `RUN`) agrega una. Docker guarda cada capa y la reutiliza mientras sus insumos no cambien; en cuanto una cambia, se vuelven a construir esa y todas las que vienen después. Por eso conviene ordenar de lo que menos cambia (el sistema, las dependencias) a lo que más cambia (el código).

```mermaid
flowchart TD
    A["FROM python:3.13-slim<br/>cambia casi nunca"] --> B["COPY pyproject.toml, uv.lock<br/>cambia al agregar una librería"]
    B --> C["RUN uv sync<br/>se reutiliza si B no cambió"]
    C --> D["COPY src<br/>cambia en cada edición del código"]
```

Un **volumen** hace que lo que está en una carpeta viva fuera de la imagen. Puede ser una carpeta del computador montada dentro del contenedor (un *bind mount*, lo que haremos con `models/`) o un espacio que administra el propio Docker. El archivo **`.dockerignore`** lista qué excluir del **contexto de construcción**, la carpeta que Docker envía para construir la imagen.
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Las instrucciones de un `Dockerfile`</strong>

Un `Dockerfile` es una lista de instrucciones que Docker ejecuta de arriba hacia abajo para construir la imagen. Estas son las que usaremos:

| Instrucción | Qué hace | En nuestro `Dockerfile` |
| --- | --- | --- |
| `FROM` | Parte de una imagen base: es la primera capa. | `python:3.13-slim`, con una versión explícita |
| `COPY` | Copia archivos del contexto de construcción a la imagen. Con `--from=` copia desde otra imagen. | `pyproject.toml` y `uv.lock`, luego `src`; el programa `uv` desde su imagen oficial |
| `WORKDIR` | Fija la carpeta de trabajo de las instrucciones siguientes y del contenedor. | `/app` |
| `ENV` | Define variables de entorno que quedan dentro de la imagen. | `UV_NO_CACHE=1` y el `PATH` del entorno virtual |
| `RUN` | Ejecuta un comando **al construir**; su resultado queda como una capa. | `uv sync` instala las dependencias; `useradd` crea el usuario |
| `USER` | Define con qué usuario corren las instrucciones siguientes y el contenedor. | un usuario sin privilegios |
| `EXPOSE` | Documenta qué puertos usa el servicio. **No los publica**: eso se hace al ejecutar. | 8000 (API) y 8501 (interfaz) |
| `HEALTHCHECK` | Un comando que Docker ejecuta cada cierto tiempo para saber si el contenedor está sano. | consulta la ruta `/health` |
| `CMD` | El comando que se ejecuta **al iniciar** el contenedor. | `uvicorn` y `streamlit` |

La diferencia entre `RUN` y `CMD` se confunde fácil: `RUN` corre una vez, al construir, y deja su resultado en la imagen (instalar paquetes); `CMD` corre cada vez que se inicia un contenedor (levantar el servicio).
</div>

Es un contenedor con dos procesos: `uvicorn` sirve la API y `streamlit` sirve la interfaz, que viene escrita en `src/aguaapr/ui.py`. El `.dockerignore` debe excluir `.env`, `.venv` y `models`, además de `data`, `notebooks`, `tests`, `scripts` y las cachés.

### 3.2 — Correr el contenedor (0,4 puntos)

Una imagen no hace nada hasta que se ejecuta como contenedor, y `docker run` recibe por banderas todo lo que la imagen no trae: los puertos que se abren hacia el computador, las variables de entorno y las carpetas que se montan.

| Bandera | Qué hace |
| --- | --- |
| `-d` | Corre el contenedor en segundo plano (*detached*). |
| `--rm` | Borra el contenedor cuando se detiene, para que no se acumulen. |
| `--name aguaapr` | Le da un nombre, para poder detenerlo y mirar sus registros. |
| `-p 8000:8000` y `-p 8501:8501` | **Publican** un puerto, como `puerto_del_computador:puerto_del_contenedor`: aquí, el de la API y el de la interfaz. Sin esto, el servicio no se alcanza desde fuera del contenedor. |
| `--env-file .env` | Pasa al contenedor las variables del archivo. |
| `-e APR_MODEL_DIR=/app/models` | Fija una variable de entorno suelta. |
| `-v "${PWD}/models:/app/models:ro"` | Monta una carpeta del computador dentro del contenedor, de solo lectura (`ro`): el modelo vive fuera de la imagen. `${PWD}` es la carpeta actual, en bash y en PowerShell. |

Ejecuten el servicio y comprueben su contrato:

```bash
docker run -d --rm --name aguaapr -p 8000:8000 -p 8501:8501 --env-file .env -e APR_MODEL_DIR=/app/models -v "${PWD}/models:/app/models:ro" aguaapr:1.0
uv run python scripts/smoke_test.py
```

Cuando terminen, `docker stop aguaapr` detiene el contenedor y, por el `--rm`, lo borra.

Imaginen que en vez de comprar una cocina industrial para un evento, la arriendan por hora: alguien más la mantiene, ustedes llevan el paquete listo y pagan solo por el tiempo que la usan. La **nube** funciona así con los computadores. Un **proveedor** (aquí, Microsoft con **Azure**) administra las máquinas, y ustedes entregan un contenedor.

**Azure Container Apps** es el servicio de Azure que ejecuta contenedores sin que haya que administrar máquinas. Estos son los conceptos que usaremos:

| Concepto de Azure | Qué es | En la cocina |
| --- | --- | --- |
| **Suscripción** | La cuenta que paga. La de estudiantes trae 100 USD de crédito. | La cuenta con que se paga el arriendo |
| **Grupo de recursos** | Una carpeta que reúne todo lo que se crea para un proyecto. Borrarlo borra todo lo que contiene. | La caja donde va todo lo del arriendo: al devolverla, se termina el contrato |

| **Entorno** | El espacio compartido donde corren una o más aplicaciones. Lo crean ustedes antes de la aplicación. | El local donde están las cocinas |
| **Aplicación** | El contenedor en ejecución, con su configuración. | La cocina funcionando |
| **Ingress** | La entrada que hace pública la aplicación: una URL con HTTPS que Azure genera. | La puerta a la calle |
| **Secreto** | Un valor sensible que la plataforma guarda aparte y la aplicación solo referencia. | La caja fuerte |
| **Réplica** | Cada copia en ejecución del contenedor. | Cada cocinero |

La plataforma, y no ustedes, decide cuántas réplicas hay, dentro de los límites que fijen.

  <strong style="display:block; margin-bottom:8px;">📖 Escalar a cero, arranque en frío y serverless</strong>

Si nadie usa la aplicación, mantener una réplica encendida es gastar sin motivo. Con un **mínimo de réplicas igual a cero**, la plataforma apaga la última réplica después de un tiempo sin tráfico, y no se paga por el tiempo apagado. Este modelo, en que no se administra una máquina y la plataforma ejecuta el código cuando llega una solicitud, se llama **serverless**.

Tiene un costo que se paga en la primera solicitud después de un rato sin tráfico: la plataforma debe arrancar un contenedor nuevo y el servicio debe cargar su modelo antes de responder. Ese retraso es el **arranque en frío**. Los límites del modelo serverless son concretos:

- **Arranque en frío**: la primera solicitud tras un periodo inactivo tarda más que las siguientes.
- **Duración máxima**: una ejecución tiene un tiempo límite y se corta al alcanzarlo.
- **Sin disco propio ni estado entre ejecuciones**: lo que deba persistir va a un servicio externo, y lo que el servicio necesite al arrancar, como el modelo, tiene que viajar con la imagen.

### 4.3 — Desplegar en Azure (1,3 puntos)

Todo lo que sigue se hace en el portal de Azure, salvo construir y subir la imagen, que se hace con Docker desde la carpeta del laboratorio, donde está el `Dockerfile`. Usen los nombres indicados (`rg-aguaapr`, `aguaapr-env` y `aguaapr`), salvo `REGION` (la región que eligieron) y `NOMBRE`, el nombre de su registro de contenedores: debe ser **único en todo Azure**, con solo letras minúsculas y números, entre 5 y 50 caracteres (por ejemplo, `aguaapr` más sus iniciales y cuatro dígitos).

Azure puede construir la imagen por ustedes con un servicio llamado ACR Tasks, pero las suscripciones de Azure for Students lo tienen bloqueado (`TasksOperationsNotAllowed`). Por eso la construirán en su computador, con el Docker que instalaron en la etapa 0, y la subirán ustedes al registro.

**Paso 1: el grupo de recursos.** Es la carpeta que reunirá todo lo que se cree, y la que se borrará cuando el equipo docente lo indique. En el portal, busquen **Grupos de recursos** (*Resource groups*), elijan **Crear** (*Create*) y completen el nombre `rg-aguaapr` y la región `REGION`.

**Paso 2: el registro de contenedores.** Es la bodega donde Azure guardará su imagen. Escriban `container registries` en la barra de búsqueda: el servicio aparece con ese nombre en inglés, aunque el resto del portal esté en español. Elijan **Create** y completen el grupo de recursos `rg-aguaapr`, el nombre `NOMBRE`, la región `REGION` y el plan (*Pricing plan*) **Basic**. Cuando termine de crearse, abran el registro y vayan a **Settings → Access keys**: activen **Admin user**, que habilita un usuario y una clave, y anoten ambos. Con ellos Docker puede subir la imagen y la aplicación puede descargarla.

**Paso 3: construir y subir la imagen.** La imagen lleva el nombre del registro como prefijo, y esa etiqueta le dice a `docker push` dónde subirla. `docker login` pide el usuario y la clave del paso anterior; al escribir la clave, la terminal no muestra nada. El `--platform` es obligatorio en la práctica: los servidores de Azure son Linux x86, y en un computador con otra arquitectura (como un Mac con chip Apple) Docker construiría por defecto una imagen que ahí no arranca.

```bash
docker login NOMBRE.azurecr.io
docker build --platform linux/amd64 -t NOMBRE.azurecr.io/aguaapr:1.0 .
docker push NOMBRE.azurecr.io/aguaapr:1.0
```

En el portal, **Services → Repositories** del registro debe mostrar el repositorio `aguaapr` con la etiqueta `1.0`.

**Paso 4: el entorno.** Es el espacio donde correrá la aplicación. Créenlo aparte, antes que la aplicación: busquen **Container Apps Environments** (escríbanlo en la barra de búsqueda), elijan **Create** y completen el grupo de recursos `rg-aguaapr`, el nombre `aguaapr-env` y la región `REGION`. Dejen el plan por defecto (**Consumption**) y el destino de registros por defecto (**Azure Log Analytics**).

**Paso 5: la aplicación.** Busquen **Container Apps**, elijan **Create → Container App** y completen:

- **Basics**: el grupo de recursos `rg-aguaapr`, el nombre `aguaapr` y, en **Container Apps environment**, el entorno `aguaapr-env` que crearon (no elijan crear uno nuevo desde aquí).
- **Container**: desmarquen **Use quickstart image**; en **Image source** elijan **Azure Container Registry** y después su registro `NOMBRE`, la imagen `aguaapr` y la etiqueta `1.0`.
- **Ingress**: marquen **Enabled**, en el tráfico elijan **Accepting traffic from anywhere** y en **Target port** escriban `8501`.

El ingress hace pública la aplicación con una URL con HTTPS que Azure genera. El `8501` es el puerto de la **interfaz**: la API (puerto 8000) queda dentro del contenedor, y solo la interfaz la consulta. Elijan **Review + create** y **Create**. Tarda unos minutos. No agreguen todavía variables de entorno: el paso 6 usa la falla de esta primera versión para comprobar lo que predijeron en 4.2.

<!-- WARNING -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #d29922; background:rgba(210,153,34,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">⚠️ El puerto por defecto no es el de la interfaz</strong>

El portal propone `80` como puerto de destino y, al activar el ingress, agrega por su cuenta tres sondas de salud (*Liveness*, *Readiness* y *Startup*) que consultan ese mismo puerto. Nuestro contenedor no escucha en el 80, sino en el 8000 (API) y el 8501 (interfaz), así que las sondas fallan: la réplica nunca queda lista, la plataforma la reinicia cada pocos minutos y la URL no termina de cargar o responde `{"error":"App port is not ready"}`. Los registros no muestran ningún error de la aplicación, porque arranca bien en cada reinicio.

Las sondas conservan el puerto que tenían cuando se creó la aplicación: cambiar después el puerto del ingress no las actualiza. Por eso el paso 7 las revisa.
</div>

**Paso 6: mirar qué pasó.** Esta primera versión no tiene `APR_API_KEY` ni `APR_GRUPO`. Abran la aplicación y, en el menú lateral, vayan a **Monitoring → Log stream** y elijan **Console**. Deberían ver el error de configuración que predijeron en 4.2: el servicio se niega a iniciar sin su configuración, y el `Field required` nombra `api_key` y `grupo`.

**Paso 7: la clave como secreto, el grupo, las sondas y las réplicas.** La clave no va en la imagen ni se escribe como variable en claro. Se guarda como **secreto** de la aplicación, y la variable de entorno solo lo referencia. Usen una clave desechable, distinta de las que usan en otros lugares.

1. En la aplicación, vayan a **Security → Secrets** y elijan **Add**. Escriban el nombre `apikey`, en **Type** elijan **Container Apps Secret** y en **Value** su clave. Guardar un secreto no crea una revisión nueva.
2. Vayan a **Application → Revisions and replicas**, elijan **Create new revision**, seleccionen el contenedor `aguaapr` y elijan **Edit**.
3. En **Environment variables**, elijan **Add**: nombre `APR_API_KEY`, **Source** igual a **Reference a secret** y, como valor, el secreto `apikey`. Agreguen otra variable: nombre `APR_GRUPO`, **Source** igual a **Manual entry** y, como valor, los nombres de quienes integran el equipo, por ejemplo `Juanito y Pepito`. Esta no es un secreto, pero el servicio la exige igual que la clave: sin ella no inicia. La interfaz la muestra bajo el título, para que los ayudantes sepan a qué equipo corresponde la URL.
4. En **Health probes**, comprueben que **Liveness**, **Readiness** y **Startup** consultan el puerto `8501`. Si alguna muestra otro, cámbienlo.
5. Guarden el contenedor con **Save**. Luego, en la pestaña **Scale**, fijen el mínimo de réplicas en `0` y el máximo en `1`, y elijan **Create**.

La API y la interfaz leen la misma variable, así que la interfaz puede consultar la API sin que nadie escriba la clave en una pantalla. La revisión nueva tarda un par de minutos en quedar activa.

**Paso 8: la URL y la verificación.** En la página principal de la aplicación (**Overview**), copien la **Application Url** y ejecuten el script de humo en su computador, ahora con `--ui`, que comprueba que la interfaz esté viva:

```bash
uv run python scripts/smoke_test.py --ui --url https://URL-DE-LA-APLICACION
```

Después **abran la URL en el navegador** y hagan una predicción con el formulario. Como la API no es pública, solo la interfaz se puede comprobar desde fuera.

<!-- TIP -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #2f81f7; background:rgba(47,129,247,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">💡 Si algo falla</strong>

Busquen el mensaje o el síntoma que tienen a la vista:

| Mensaje o síntoma | Qué significa | Qué hacer |
| --- | --- | --- |
| `MissingSubscriptionRegistration` | Falta activar un proveedor en la suscripción. | En **Suscripciones**, abran la suya, vayan a **Settings → Resource providers**, busquen el que el mensaje nombre (por ejemplo, `Microsoft.App`), elijan **Register**, esperen a que su estado sea **Registered** y repitan el paso. |
| `RequestDisallowedByAzure` | Esa región no acepta el recurso en su suscripción. | Prueben con otra región y vuelvan a crear en ella lo que ya habían creado en la región anterior y lo que falte. Si ninguna funciona, avisen al equipo docente. |
| `ExpressAzureMonitorLogDestNotSupported` | La aplicación intentó usar un entorno de tipo Express, que no admite ese destino de registros. | Creen el entorno aparte, como en el paso 4, y elíjanlo en el paso 5. |
| La URL no termina de cargar, o dice `App port is not ready` | El puerto del ingress o de las sondas no es el de la interfaz. | Revisen que sea el `8501` en el ingress y en las sondas (paso 7). |
| `Authentication failed when pulling container image` | La aplicación no pudo descargar la imagen del registro. | Comprueben que **Admin user** esté activo en **Access keys** del registro. |
| `unauthorized` al ejecutar `docker push` | Docker no está autenticado en el registro. | Repitan `docker login`. |
| `exec format error` | La imagen se construyó para otra arquitectura. | Reconstruyan con `--platform linux/amd64`. |
| La URL abre la interfaz, pero la predicción falla con un error de conexión | La API no está en ejecución dentro del contenedor. | Miren el **Log stream**. |

Si nada de esto explica el error, avisen al equipo docente en vez de crear recursos en otra suscripción.
</div>

<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #2da44e; background:rgba(45,164,78,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">🧪 Comprobación manual</strong>

`smoke_test.py --ui --url https://...` imprime sus dos líneas con ✓, y la interfaz abierta en el navegador muestra bajo el título `Grupo:` seguido de los nombres de su equipo y entrega una predicción. Si la URL no termina de cargar, miren el **Log stream**: un `Field required` que nombra `grupo` o `api_key` indica que falta esa variable (paso 7).
